```python
import pandas as pd

from sklearn.preprocessing import OneHotEncoder

import torch.nn as nn
```

```python
df = pd.read_csv("all_pull_requests_with_update_over_time.csv")
df
```

```python
#remove unnecessary columns
df.drop("pr_title",inplace = True,axis=1)
df.drop("class_names",inplace = True,axis=1)
df.drop("commit_sha",inplace = True,axis=1)
df.drop("commit_message",inplace = True,axis=1)
df.drop("file_path",inplace = True,axis=1)
df.drop("additions",inplace = True,axis=1)
df.drop("deletions",inplace = True,axis=1)
#df.drop("patch",inplace = True,axis=1)

#remove duplicates
#df = df.drop_duplicates(keep='first')

df
```

```python
# Filter rows where 'labels' doesn't contain '|'
df = df[~df['labels'].str.contains('|', regex=False)]

df
```

```python
# OneHotEncoder creation
encoder = OneHotEncoder()
columns_to_encode = ['labels', 'file_name']

# convert chosen columns and create new dataframe
encoded_data = pd.DataFrame(encoder.fit_transform(df[columns_to_encode]).toarray())

# determine new column names
encoded_data.columns = encoder.get_feature_names_out(columns_to_encode)

# Add new data columns to dataframe
df = pd.concat([df.reset_index(drop=True), encoded_data.reset_index(drop=True)], axis=1)

df
```

```python
# convert created_at and closed_at columns to datetime format
df['created_at'] = pd.to_datetime(df['created_at'])
df['closed_at'] = pd.to_datetime(df['closed_at'])

# convert created_at and closed_at columns to epoch-time format
df['created_at_epoch'] = df['created_at'].astype('int64') // 10**9
df['closed_at_epoch'] = df['closed_at'].astype('int64') // 10**9
df
```
